## Le regole per usare Gemini nel corso

1. **Leggete sempre il codice** prima di eseguirlo. Se una riga non vi è chiara, chiedete di spiegarla.
2. **Eseguitelo e controllate il risultato**: è plausibile? Torna con quello che sapete?
3. **Gemini può sbagliare con sicurezza**: inventare funzioni o colonne che non esistono, scegliere la media invece della mediana. Il codice che gira non è per forza giusto.
4. **Prima provate a capire cosa vi serve**: Gemini è uno strumento, non il pilota. Nei task trovate cosa cercare, e cercare è metà del mestiere.
5. **Non incollate dati personali o riservati** nelle richieste.

## Parte 2 — Lo slicing

Prima di tornare al dataset, allenati sui tagli con qualcosa di piccolo. Ricorda: in `[inizio:fine]` i numeri sono **tagli tra le caselle**, si parte da 0, e la fine è esclusa.

### 6. Tagliare un testo

Data la variabile `codice` qui sotto, estrai solo il **prefisso** (`"IMP"`) e solo il **numero** (`"0013"`), scrivendo sia l'inizio sia la fine.

🔎 *Da cercare:* `python slicing stringhe`

```
codice = "IMP0013"
print(codice[0:3])
print(codice[3:7])
```

### 7. Le omissioni

Se l'inizio è 0 puoi ometterlo; se la fine è la fine puoi ometterla.

1. Riscrivi le due estrazioni del passo precedente **omettendo** un numero in ciascuna.
2. Verifica con `==` che il risultato sia identico a quello di prima.
3. Cosa dà `codice[:]`?

🔎 *Da cercare:* `python slicing omettere inizio fine`

```
print(codice[:3], codice[3:])
print(codice[:3] == codice[0:3], codice[3:] == codice[3:7])
print(codice[:])
```

*Risposta:* `codice[:3]` equivale a `codice[0:3]`; `codice[3:]` arriva fino alla fine. `codice[:]` è tutto il testo.

### 8. Il passo

Lo slicing ha un terzo numero: `[inizio:fine:passo]`. Con la variabile `lettere` qui sotto, ottieni:

1. una lettera ogni due, partendo dalla A: `"ACEG"`
2. una lettera ogni due, partendo dalla B: `"BDFH"`
3. tutte le lettere al contrario: `"HGFEDCBA"`

🔎 *Da cercare:* `python slicing step` · `python invertire stringa`

```
lettere = "ABCDEFGH"
print(lettere[::2])
print(lettere[1::2])
print(lettere[::-1])
```

## Parte 3 — Selezionare righe e colonne

### 9. Una fetta del dataset

Mostra solo le colonne `id_impresa`, `anno` e `fatturato`, e solo le righe dalla **100 alla 109 comprese**. Controlla: quante righe vedi? Devono essere 10.

🔎 *Da cercare:* `pandas selezionare più colonne`

```
df.iloc[100:110][["id_impresa", "anno", "fatturato"]]
```

*Risposta:* Il secondo numero è escluso: per arrivare alla 109 bisogna scrivere 110. Per più colonne per nome servono le doppie parentesi: una lista di nomi dentro `df[...]`.

### 10. Una sola impresa

Mostra solo le righe dell'impresa `IMP0001`. Quante sono? Quante te ne aspetti? Poi prova con `IMP0013`.

🔎 *Da cercare:* `pandas filtrare righe condizione`

```
print(len(df[df["id_impresa"] == "IMP0001"]))
print(len(df[df["id_impresa"] == "IMP0013"]))
df[df["id_impresa"] == "IMP0013"]
```

*Risposta:* `IMP0001` ha 5 righe, una per anno. `IMP0013` ne ha 6: un anno compare due volte. Primo indizio dei duplicati.

Se ogni impresa compare una volta per ogni anno, quante righe dovrebbe avere il dataset? Fai il calcolo con Python e confrontalo con il passo 3.

### 12. Il fatturato

Calcola il fatturato **medio** e il fatturato **massimo**. Funziona? Il risultato ha senso?

🔎 *Da cercare:* `pandas mean` · `pandas max` · l'ultima riga dell'eventuale errore

```
# df["fatturato"].mean()  -> errore: non si può fare la media di un testo
df["fatturato"].max()
```

*Risposta:* La media dà errore perché la colonna è testo. Il massimo invece funziona, ma restituisce `'n.d.'`: confronta i testi in ordine alfabetico, non i numeri. Nessun errore, risultato senza senso.

*Risposta:* La stessa regione è scritta in modi diversi: `Lombardia`, `lombardia`, `LOMBARDIA`, `Lombardia ` (con spazio), `Emilia-Romagna` ed `Emilia Romagna`, `Veneto` e `Veneto `. Per Python sono valori diversi.

### Qualche aiuto per traccia

Alcune tracce chiedono cose che nel corso non abbiamo fatto. Ecco da dove partire.

- **Crescita tra due anni** (ESG, innovazione): mettete le imprese sulle righe e gli anni sulle colonne (cerca `pandas pivot_table`, bonus B2 della L4), poi $\text{crescita}\,\% = \left(\frac{\text{fatturato}_{2023}}{\text{fatturato}_{2019}} - 1\right) \times 100$.
- **Andamento nel tempo** (recensioni): dalla colonna `data` estraete l'anno o il mese (cerca `pandas dt year` · `pandas dt to_period M`), poi `groupby`.
- **Esperimento**: prima confrontate le medie dei due gruppi (`groupby`). Per sapere quanto siete sicuri della differenza usate la regressione della L7 con la condizione come variabile: `smf.ols("intenzione_acquisto ~ condizione", data=...)`. Con una variabile di testo, statsmodels crea da solo la variabile 0/1: il coefficiente è la differenza tra i due messaggi, e le colonne `[0.025 0.975]` del riepilogo sono l'**intervallo di confidenza** (cerca `statsmodels conf_int`). Per "per chi?": rifate il confronto separatamente per giovani e meno giovani, o per clienti abituali e no.

## Come caricare i dati

La cella qui sotto mostra come caricare ciascun dataset del corso. Tenete solo le righe che vi servono e cancellate le altre.

```
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf

URL = "https://raw.githubusercontent.com/andrealorenzon/SIMA2026/main/dati"

# imprese e punteggi ESG (E2)
imprese = pd.read_excel(f"{URL}/imprese_pulito.xlsx")
esg = pd.read_excel(f"{URL}/esg.xlsx")

# esperimento
esperimento = pd.read_excel(f"{URL}/esperimento.xlsx")

# recensioni: un file per marca, nella repository del corso (E3)
![ -d SIMA2026 ] || git clone -q --depth 1 https://github.com/andrealorenzon/SIMA2026.git
from pathlib import Path
pezzi = []
for f in sorted(Path("SIMA2026/dati/recensioni").glob("*.xlsx")):
    d = pd.read_excel(f)
    d["marca"] = f.stem
    pezzi.append(d)
recensioni = pd.concat(pezzi)

print(imprese.shape, esg.shape, esperimento.shape, recensioni.shape)
```

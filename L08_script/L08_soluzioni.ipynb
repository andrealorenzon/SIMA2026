## Contesto

Il notebook è perfetto per esplorare: provi, guardi, correggi. Ma ha un difetto: le celle si possono eseguire in qualsiasi ordine, e il risultato dipende da cosa hai eseguito prima. Un'analisi che deve essere **rifatta** (da te tra sei mesi, da un collega, da un revisore) va messa in uno **script**: un file `.py` che si esegue sempre dall'inizio alla fine, nello stesso ordine.

## Parte 2 — Il tuo primo script

In Colab una cella che inizia con `%%writefile nome.py` non esegue il codice: lo **salva in un file**. Con `!python nome.py` lo esegui, dall'inizio alla fine, in una sessione pulita.

### 2. Scrivi lo script

Completa lo scheletro qui sotto: carica il dataset, calcola per ogni settore il numero di imprese diverse, il fatturato mediano e gli addetti mediani, ordina per fatturato mediano, salva in `riepilogo_settori.xlsx` e stampa il risultato. Compila anche l'intestazione: chi, quando, quali dati, cosa fa. Poi esegui la cella (crea il file, non mostra risultati).

🔎 *Da cercare:* `colab writefile` · `pandas groupby agg` · `pandas to_excel`

```
%%writefile analisi.py
"""Riepilogo delle imprese per settore.

Autore: Mario Rossi
Data: ottobre 2026
Dati: imprese_pulito.xlsx (corso SIMA 2026)
Cosa fa: per ogni settore calcola numero di imprese, fatturato mediano e addetti
mediani, e salva il risultato in riepilogo_settori.xlsx.
"""
import pandas as pd

URL = "https://raw.githubusercontent.com/andrealorenzon/SIMA2026/main/dati/imprese_pulito.xlsx"

# 1. carica i dati
df = pd.read_excel(URL)

# 2. calcola il riepilogo per settore
riepilogo = df.groupby("settore").agg(
    imprese=("id_impresa", "nunique"),
    fatturato_mediano=("fatturato", "median"),
    addetti_mediani=("addetti", "median"),
).sort_values("fatturato_mediano", ascending=False)

# 3. salva e stampa
riepilogo.to_excel("riepilogo_settori.xlsx")
print(riepilogo)
```

## Parte 3 — Leggere il codice degli altri

### 5. Lo script di Gemini

Un collega ha chiesto a Gemini lo script qui sotto. Il commento iniziale dice cosa **dovrebbe** fare. Eseguilo, poi leggilo riga per riga: fa davvero quello che dice? Trova **due** differenze tra il commento e il codice, e correggile.

🔎 *Da cercare:* `python operatori di confronto` · `pandas median mean`

```
%%writefile script_gemini.py
"""Fatturato MEDIANO per settore, considerando solo le imprese
con ALMENO 10 addetti, nell'anno 2022."""
import pandas as pd

df = pd.read_excel("https://raw.githubusercontent.com/andrealorenzon/SIMA2026/main/dati/imprese_pulito.xlsx")
df = df[df["anno"] == 2022]
df = df[df["addetti"] >= 10]
print(df.groupby("settore")["fatturato"].median().round(1))

# poi: !python script_gemini.py
```

```
!python script_gemini.py
```

*Risposta:* 1) "almeno 10 addetti" vuol dire `>= 10`, ma il codice usa `> 10` ed esclude le imprese con esattamente 10 addetti. 2) Il commento dice fatturato **mediano**, il codice calcola la **media**. Lo script gira senza errori e i numeri sembrano plausibili: solo leggendolo si trovano gli sbagli.

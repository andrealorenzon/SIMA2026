## Contesto

Marco ha **5.000 €** e li investe al **2,5% annuo**. Vogliamo sapere quanto avrà nel tempo.

### 1. Le variabili

Crea una variabile `capitale` con valore 5000 e una variabile `tasso` con valore 2,5% (scritto come numero decimale).

🔎 *Da cercare:* `python variabili · python numeri decimali`

```
capitale = 5000
tasso = 0.025
```

### 2. Dopo un anno

Calcola quanto avrà Marco dopo un anno e stampa il risultato.

🔎 *Da cercare:* `python print · python operazioni aritmetiche`

```
print(capitale * (1 + tasso))
```

### 3. Dopo 10 anni

Calcola quanto avrà dopo 10 anni. Formula: $\text{capitale} \times (1 + \text{tasso})^{\text{anni}}$. Salva il risultato in una variabile `montante`.

🔎 *Da cercare:* `python potenza`

```
anni = 10
montante = capitale * (1 + tasso) ** anni
print(montante)
```

### 4. Arrotondare

Arrotonda `montante` a 2 decimali.

🔎 *Da cercare:* `python round`

```
montante = round(montante, 2)
print(montante)
```

### 5. Una frase completa

Stampa la frase: *Dopo 10 anni Marco avrà X euro*, dove X è il valore calcolato.

🔎 *Da cercare:* `python f-string`

```
print(f"Dopo {anni} anni Marco avrà {montante} euro")
```

### 6. Vero o falso?

Rispondi con Python, senza fare conti a mano:

1. Dopo 10 anni Marco avrà **più di** 6.000 €?
2. Avrà **esattamente** 6400.42 €?
3. Avrà **esattamente** 6400 €?

Che tipo di dato ottieni come risposta?

🔎 *Da cercare:* `python operatori di confronto` · `python type`

```
print(montante > 6000)
print(montante == 6400.42)
print(montante == 6400)
print(type(montante > 6000))
```

**Attenzione:** cosa succede se per sbaglio scrivi `montante = 6400` invece di `montante == 6400`? Provalo nella cella qui sotto, poi stampa `montante`. Alla fine riesegui le celle dei passi 3 e 4 per rimettere `montante` a posto.

```
montante = 6400
print(montante)
```

*La mia spiegazione:* Nessun errore: `=` non chiede niente, **mette** 6400 dentro `montante`, sovrascrivendo il valore calcolato. `==` invece fa una domanda e risponde `True` o `False` (tipo `bool`).

```
anni = 10
montante = round(capitale * (1 + tasso) ** anni, 2)
```

Esegui questa cella **senza modificarla**. Cosa è successo? Scrivi la tua spiegazione nella cella di testo sotto.

### B1. Anno per anno

Stampa una riga per ogni anno da 1 a 10, con il capitale di Marco a fine anno.

🔎 *Da cercare:* `python for range`

```
for a in range(1, 11):
    print(a, round(capitale * (1 + tasso) ** a, 2))
```

### B2. Quanti anni per raddoppiare?

Dopo quanti anni Marco avrà almeno 10.000 €?

🔎 *Da cercare:* `python while`

```
valore, a = capitale, 0
while valore < 2 * capitale:
    valore = valore * (1 + tasso)
    a = a + 1
print(a)
```

## 0. Preparazione

Esegui questa cella così com'è: carica la libreria **pandas** e salva l'indirizzo del file nella variabile `url`.

```
import pandas as pd

url = "https://raw.githubusercontent.com/andrealorenzon/SIMA2026/main/dati/imprese.xlsx"
```

### 1. Carica il file

Leggi il file Excel che si trova all'indirizzo `url` e salvalo in una variabile chiamata `df`.

🔎 *Da cercare:* `pandas read_excel`

```
df = pd.read_excel(url)
```

### 2. Dai un'occhiata

Mostra le prime 10 righe di `df`.

🔎 *Da cercare:* `pandas head`

```
df.head(10)
```

### 3. Quanto è grande?

Quante righe e quante colonne ha `df`?

🔎 *Da cercare:* `pandas shape`

```
df.shape
```

### 4. Che tipo di dati?

Mostra il tipo di dato di ogni colonna. Guarda bene: c'è una colonna che dovrebbe essere un numero ma non lo è?

🔎 *Da cercare:* `pandas dtypes`

```
df.dtypes
```

*Risposta:* `fatturato` è di tipo testo (`object` o `str`), non numerico.

### 5. Quante imprese, quanti anni?

Quante imprese **diverse** ci sono nella colonna `id_impresa`? E quanti anni diversi nella colonna `anno`?

🔎 *Da cercare:* `pandas nunique · pandas selezionare una colonna`

```
print(df["id_impresa"].nunique())
print(df["anno"].nunique())
```

### 6. I conti tornano?

Se ogni impresa compare una volta per ogni anno, quante righe dovrebbe avere il dataset? Fai il calcolo con Python e confrontalo con il passo 3.

```
500 * 5
```

*Risposta:* Dovrebbero essere 2.500 righe, ma sono 2.515: ci sono 15 righe in più. Probabilmente righe duplicate (lo verifichiamo nella L3).

### 7. Il fatturato

Calcola il fatturato **medio** e il fatturato **massimo**. Funziona? Il risultato ha senso?

🔎 *Da cercare:* `pandas mean` · `pandas max` · l'ultima riga dell'eventuale errore

```
# df["fatturato"].mean()  -> errore: non si può fare la media di un testo
df["fatturato"].max()
```

*Risposta:* La media dà errore perché la colonna è testo. Il massimo invece funziona, ma restituisce `'n.d.'`: confronta i testi in ordine alfabetico, non i numeri. Nessun errore, risultato senza senso.

### 8. Le regioni

Quante righe ci sono per ogni regione? Guarda la lista con attenzione: c'è qualcosa di strano?

🔎 *Da cercare:* `pandas value_counts`

```
df["regione"].value_counts()
```

*Risposta:* La stessa regione è scritta in modi diversi: `Lombardia`, `lombardia`, `LOMBARDIA`, `Lombardia ` (con spazio), `Emilia-Romagna` ed `Emilia Romagna`, `Veneto` e `Veneto `. Per Python sono valori diversi.
